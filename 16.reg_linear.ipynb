```python
import pandas as pd
import numpy as np

# aprendizado
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import r2_score

import matplotlib.pyplot as plt
```

```python
try:
    df = pd.read_csv('03.BaseDPEvolucaoMensalCisp.csv',
                     encoding='iso-8859-1', sep=';')
    df.info()
except Exception as e:
    print(f'Erro ao obter dados: {e}')
```

```python
coluna1 = 'lesao_corp_culposa'
coluna2 = 'lesao_corp_dolosa'
```

```python
# antes de usar Regressão Linear vamos testar a correlação 
r = df[coluna1].corr(df[coluna2])
print(f'r de Pearson: {r:.4f}')
```

```python
# colocar as colunas escolhidas nas variaveis
# dependendo dos dados não é necessario, aqui iremos 
# 'copiar' para não ter chance de alterar dados reais
df2 = df[[coluna1, coluna2]].dropna()

X = df2[[coluna1]]
y = df2[coluna2]

# Normalizar a escala
scaler = MinMaxScaler()
X_scaled = X #scaler.fit_transform(X)
```

```python
# dividir os dados, treino - teste
X_treino, X_teste, y_treino, y_teste = (
    train_test_split(X_scaled, y, test_size=0.2,
                     random_state=42)
)
```

```python
# treinar modelo
model = LinearRegression()
model.fit(X_treino, y_treino)
```

```python
# Testar e Verificar o modelo
y_pred = model.predict(X_teste)

r2 = r2_score(y_teste, y_pred)

print(f'r2_score: {r2:.4f}')
print(f'Coeficiente angular (a): {model.coef_[0]:.4f}')
print(f'Intercepto (b): {model.intercept_:.4f}')
```

```python
# novo_dado = scaler.transform([[3]])
novo_dado = 200
previsao = model.predict([[novo_dado]])
print(f'Para {novo_dado} {coluna1} -> {coluna2} previsto: {previsao[0]:.2f}')
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(X_teste, y_teste, alpha=0.7, label='Dados reais')
plt.plot(X_teste, y_pred, linewidth=2, color='red', label='Reta de regressão')
plt.xlabel(f'{coluna1} (normalizado)')
plt.ylabel(f'{coluna2}')
plt.title(f'Regressão Linear | r={r2:.4f}')
plt.legend()
plt.tight_layout()
plt.show()
```
