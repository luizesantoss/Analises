```python
import polars as pl
from sqlalchemy import create_engine  # create engine cria conexao c o banco
import pyarrow as pa
```

```python
# Configuração
PARQUET_PATH = r'dados_bronze/bf.parquet'
TABELA = 'bolsa_familia'
BATCH_SIZE = 10000
ENGINE_URL = (
'mysql+pymysql://root:@localhost:3306/bolsa_familia'
)
```

```python
# Conexão
engine = create_engine(ENGINE_URL)
```

```python
# Leitura
print('Lendo arquivo parquet')
df = pl.read_parquet(PARQUET_PATH)
```

```python
# escrita em batch (lote) no banco
total = df.shape[0]
linhas = 0
```

```python


from pathlib import Path

Path('dados_silver').mkdir(parents=True, exist_ok=True)

for i, batch in enumerate(df.iter_slices(n_rows=BATCH_SIZE)):
    batch_pd = batch.to_pandas()

    modo = 'replace' if i == 0 else 'append'

    try:
        # tenta escrever no banco MySQL
        batch_pd.to_sql(
            name=TABELA,
            con=engine,
            if_exists=modo,
            index=False
        )
    except Exception as e:
        print(f"Erro ao escrever no banco MySQL: {e}. Salvando em parquet local.")
        # fallback: salva cada lote em arquivo parquet local
        batch_pd.to_parquet(f'dados_silver/{TABELA}_batch_{i+1}.parquet', index=False)

    linhas += batch_pd.shape[0]
    percent = (linhas/total) * 100
    print(f'Lote {i+1} | {percent:.2f}%')

print('Escrita Finalizada')
```
