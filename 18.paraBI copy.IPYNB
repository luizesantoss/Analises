```python
import numpy as np
import pandas as pd 
```

```python
df = pd.read_csv('03.BaseDPEvolucaoMensalCisp.csv',
                 sep=';', encoding='iso-8859-1')

df.columns = df.columns.str.strip().str.lower() \
.str.replace(" ", "_")
df = df.dropna(subset=['cisp', 'regiao', 
                       'ano', 'roubo_veiculo'])

# calcular quartis
q1 = df['roubo_veiculo'].quantile(0.25)
q3 = df['roubo_veiculo'].quantile(0.75)

# Identificar outliers
iqr = q3 - q1
lim_inf = q1 - 1.5 * iqr
lim_sup = q3 + 1.5 * iqr

df_outliers_sup = df[df['roubo_veiculo'] > lim_sup].copy()
df_outliers_inf = df[df['roubo_veiculo'] < lim_inf].copy()

df_outliers_sup['flag'] = 'outlier_sup'
df_outliers_inf['flag'] = 'outlier_inf'

resultado_outlier = pd.concat([df_outliers_inf, 
                               df_outliers_sup])
resultado_outlier = resultado_outlier[['cisp', 'regiao',
                        'ano', 'roubo_veiculo', 'flag']]
resultado_outlier.to_csv('dp_outliers.csv', sep=';',
                         encoding='utf-8-sig',
                         index=False)
```
